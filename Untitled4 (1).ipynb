```python
# Importing libraries
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer, ENGLISH_STOP_WORDS
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import re
```

```python
# Loading the dataset
beer_data = pd.read_csv("train.csv")
```

```python
beer_data
```

```python
# Data Cleaning
# Drop rows with missing 'review/text'
beer_data = beer_data.dropna(subset=['review/text'])
```

```python
beer_data.head()
```

```python
beer_data.shape
```

```python
beer_data = beer_data.copy()
beer_data['user/gender'] = beer_data['user/gender'].fillna('Unknown')
```

```python
beer_data = beer_data.fillna({'user/gender': 'Unknown'})
```

```python
# Drop columns with high missing values and low relevance
beer_data.drop(['user/birthdayRaw', 'user/birthdayUnix', 'user/ageInSeconds', 'user/profileName'], axis=1, inplace=True)
```

```python
# Text Preprocessing
def preprocess_text(text):
    text = text.lower()  # Lowercasing
    text = re.sub(r'\d+', '', text)  # Removing numbers
    text = re.sub(r'[^\w\s]', '', text)  # Removing punctuation
    text = ' '.join([word for word in text.split() if word not in ENGLISH_STOP_WORDS])  # Removing stopwords
    return text

beer_data['cleaned_review'] = beer_data['review/text'].apply(preprocess_text)
```

```python
# TF-IDF Vectorization
tfidf = TfidfVectorizer(max_features=200)
tfidf_matrix = tfidf.fit_transform(beer_data['cleaned_review'])
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=tfidf.get_feature_names_out())
```

```python
# Encoding Categorical Features
label_encoder = LabelEncoder()
beer_data['beer/style_encoded'] = label_encoder.fit_transform(beer_data['beer/style'])
beer_data['user/gender_encoded'] = label_encoder.fit_transform(beer_data['user/gender'])
```

```python
# Selecting and Downcasting Numerical Features
numerical_features = beer_data[['beer/ABV', 'review/appearance', 'review/aroma',
                                'review/palate', 'review/taste',
                                'beer/style_encoded', 'user/gender_encoded']].apply(pd.to_numeric, downcast='float')
```

```python
# Combining Features
final_features = pd.concat([numerical_features.reset_index(drop=True), tfidf_df.reset_index(drop=True)], axis=1)
target = beer_data['review/overall']
```

```python
final_features = final_features.reset_index(drop=True)
target = target.reset_index(drop=True)

# Now sample correctly
sampled_data = final_features.sample(frac=0.2, random_state=42)
sampled_target = target.loc[sampled_data.index]  # This should now work
```

```python
sampled_target = target.sample(frac=0.2, random_state=42)
sampled_data = final_features.loc[sampled_target.index]
```

```python
sampled_target = target.iloc[sampled_data.index.to_list()]
```

```python
# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(sampled_data, sampled_target, test_size=0.2,random_state = 45)
```

```python
# Model Training
# Random Forest Regressor
rf_model = RandomForestRegressor(random_state=42)
rf_model.fit(X_train, y_train)
rf_predictions = rf_model.predict(X_test)
```

```python
# Linear Regression
lr_model = LinearRegression()
lr_model.fit(X_train, y_train)
lr_predictions = lr_model.predict(X_test)
```

```python
# Model Evaluation
def evaluate_model(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    return mae, rmse, r2
```

```python
# Evaluation Metrics (Regression)
rf_mae, rf_rmse, rf_r2 = evaluate_model(y_test, rf_predictions)
lr_mae, lr_rmse, lr_r2 = evaluate_model(y_test, lr_predictions)
```

```python
# Displaying  Results
print(f"Random Forest - MAE: {rf_mae}, RMSE: {rf_rmse}, R²: {rf_r2}")
print(f"Linear Regression - MAE: {lr_mae}, RMSE: {lr_rmse}, R²: {lr_r2}")
```
